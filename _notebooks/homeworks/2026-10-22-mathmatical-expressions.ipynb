## Popcorn Hack 1

```python
const results = [
    `9 + 6 = ${9 + 6}`, // Expected value: 15
    `9 - 6 = ${9 - 6}`, // Expected value: 3
    `9 * 6 = ${9 * 6}`, // Expected value: 54
    `9 / 6 = ${9 / 6}`, // Expected value: 1.5
    `9 mod 6 = ${9 % 6}`, // Expected value: 3
    `9 ^ 2 = ${9 ** 2}` // Expected value: 81
]
console.log(results.join('\n'));

// Division and remainder do not give the same result because division is how many times one number can fit into another, while the remainder is what is left over after a whole number division. For example, when dividing 9 by 6, it fits 1 time (1 * 6 = 6) and leaves a remainder of 3 (9 - 1 * 6 = 3).
```

## Popcorn Hack 2

```python
const result = [
    `(8 + 2) * 3 = ${(8 + 2) * 3}`, // Expected value: 30
    `8 + (2 * 3) = ${8 + (2 * 3)}`, // Expected value: 14
]
console.log(result.join('\n'));
```

## Homework Hack

```python

const itemPrice = 12;
const quantity = 4;
const discountRate = 0.25;
const budget = 30;

const subtotal = itemPrice * quantity;
const discount = subtotal * discountRate;
const total = subtotal - discount;
const withinBudget = total <= budget;

const results = [
  `Subtotal: ${subtotal}`,
  `Discount: ${discount}`,
  `Total: ${total}`,
  `Within budget: ${withinBudget}`
];
console.log(results.join("\n"));
```
