## Popcorn Hack 1

```python
class Hero {
  constructor(gamerTag) {
    this.gamerTag = gamerTag;
  }
}

let hero = new Hero("  pixel knight  ");


let displayGamerTag = hero.gamerTag.trim().toUpperCase();

// The brackets [ ] let you see any leftover spaces.
console.log("Original tag:    [" + hero.gamerTag + "]");
console.log("Leaderboard tag: [" + displayGamerTag + "]");
```

## Popcorn Hack 2

```python
class Player {
  constructor(name, health) {
    this.name = name;
    this.health = health;
  }

  heal() {
    this.health += 10;
  }

  damage() {
    this.health -= 10;
  }
}

let player = new Player("Alex", 50);
player.heal();
console.log("Health after healing: " + player.health); // 60

player.damage();
console.log("Health after damage: " + player.health); // 50
```

## Homework Hack

```python
class Enemy {
    constructor(name, evilness) {
        this.name = name;
        this.evilness = evilness;
    }

    cleanUpName() {
        this.name = this.name.trim().toUpperCase();
    }

    anger() {
        this.evilness += 10;
    }

    soften() {
        this.evilness -= 10;
    }
}

let object = new Enemy("  doctor doom  ", 200);
object.cleanUpName();
console.log("Name after cleanup: " + object.name);
object.soften();
console.log("Evilness percentage after softening: " + object.evilness);
object.anger();
console.log("Evilness percentage after angering: " + object.evilness);
```
